### Transportation Data Cleaning

The transportation dataset initially contained 13,754 records.

Three exact duplicate records were identified and removed, leaving 13,751 records.

Next, records with missing crash dates or borough information were removed because these attributes are essential for the planned temporal and geographical analysis.

This resulted in 13,255 remaining records.

Borough names were standardized by removing unnecessary spaces and converting all values to uppercase. This ensures consistency with the housing dataset and allows geographical aggregation during integration.

Records with missing ZIP codes and geographical coordinates were retained because they can still contribute to borough-level analysis.

```python
# Make sure dates are in datetime format

housing_clean["received_date"] = pd.to_datetime(
    housing_clean["received_date"]
)

transport_clean["crash_date"] = pd.to_datetime(
    transport_clean["crash_date"]
)

# Keep only the overlapping period

housing_clean = housing_clean[
    (housing_clean["received_date"] >= "2022-01-01") &
    (housing_clean["received_date"] < "2026-07-01")
].copy()

transport_clean = transport_clean[
    (transport_clean["crash_date"] >= "2022-01-01") &
    (transport_clean["crash_date"] < "2026-07-01")
].copy()

print("Housing rows:", len(housing_clean))
print("Transportation rows:", len(transport_clean))
```

```python
housing_clean["year"] = (
    housing_clean["received_date"].dt.year
)

transport_clean["year"] = (
    transport_clean["crash_date"].dt.year
)

print("Housing years:")
print(housing_clean["year"].value_counts().sort_index())

print("\nTransportation years:")
print(transport_clean["year"].value_counts().sort_index())
```

```python
from urllib.parse import urlencode

housing_id = "ygpa-z7cr"

api_url = f"https://data.cityofnewyork.us/resource/{housing_id}.csv"

params = {
    "$limit": 500000,
    "$where": "received_date >= '2022-01-01T00:00:00' AND received_date < '2026-07-01T00:00:00'",
    "$order": "received_date ASC"
}

url = api_url + "?" + urlencode(params)

housing_new = pd.read_csv(url, low_memory=False)

print("New housing dataset shape:", housing_new.shape)

housing_new["received_date"] = pd.to_datetime(
    housing_new["received_date"],
    errors="coerce"
)

print("\nHousing records by year:")
print(
    housing_new["received_date"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

```python
from urllib.parse import urlencode

params = {
    "$select": (
        "borough, "
        "date_extract_y(received_date) as year, "
        "count(distinct complaint_id) as unique_complaints"
    ),
    "$where": (
        "received_date >= '2022-01-01T00:00:00' "
        "AND received_date < '2026-07-01T00:00:00' "
        "AND complaint_id IS NOT NULL"
    ),
    "$group": "borough, year",
    "$order": "year, borough"
}

url = (
    "https://data.cityofnewyork.us/resource/ygpa-z7cr.csv?"
    + urlencode(params)
)

housing_grouped = pd.read_csv(url)

print("Housing aggregation:")
display(housing_grouped)

print("\nRecords by year:")
print(housing_grouped.groupby("year")["unique_complaints"].sum())
```

```python
# Make sure severe injuries are numeric
transport_clean["total_severe_injuries"] = pd.to_numeric(
    transport_clean["total_severe_injuries"], errors="coerce"
).fillna(0)

# Group transportation data by borough and year
transport_grouped = (
    transport_clean.groupby(["crash_borough", "year"])
    ["total_severe_injuries"]
    .sum()
    .reset_index()
)

# Rename borough column so both datasets match
transport_grouped = transport_grouped.rename(
    columns={"crash_borough": "borough"}
)

print("Transportation aggregation:")
display(transport_grouped)
```

```python
# Combine both datasets using borough and year
combined_data = pd.merge(
    housing_grouped,
    transport_grouped,
    on=["borough", "year"],
    how="outer"
)

# Replace missing values with 0
combined_data = combined_data.fillna(0)

# Sort the results
combined_data = combined_data.sort_values(
    by=["year", "borough"]
).reset_index(drop=True)

print("Combined dataset:")
display(combined_data)

print("Dataset shape:", combined_data.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart 1: Housing complaints over time
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=combined_data,
    x="year",
    y="unique_complaints",
    hue="borough",
    marker="o"
)

plt.title("Housing Complaints by Borough (2022–2026)")
plt.xlabel("Year")
plt.ylabel("Number of Unique Complaints")
plt.legend(title="Borough")
plt.grid(True)
plt.show()


# Chart 2: Severe injuries over time
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=combined_data,
    x="year",
    y="total_severe_injuries",
    hue="borough",
    marker="o"
)

plt.title("Severe Transportation Injuries by Borough (2022–2026)")
plt.xlabel("Year")
plt.ylabel("Total Severe Injuries")
plt.legend(title="Borough")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=combined_data,
    x="unique_complaints",
    y="total_severe_injuries",
    hue="borough",
    style="year",
    s=120
)

plt.title("Housing Complaints vs Severe Transportation Injuries")
plt.xlabel("Number of Unique Housing Complaints")
plt.ylabel("Total Severe Transportation Injuries")
plt.grid(True)
plt.show()
```

```python
# Create a table for the heatmap
heatmap_data = combined_data.pivot(
    index="borough",
    columns="year",
    values="total_severe_injuries"
)

plt.figure(figsize=(10, 5))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd"
)

plt.title("Severe Transportation Injuries by Borough and Year")
plt.xlabel("Year")
plt.ylabel("Borough")
plt.show()
```

```python
# Descriptive statistics for our combined dataset

print("HOUSING COMPLAINTS SUMMARY")
display(combined_data["unique_complaints"].describe())

print("\nSEVERE TRANSPORTATION INJURIES SUMMARY")
display(combined_data["total_severe_injuries"].describe())

print("\nAVERAGE VALUES BY BOROUGH")
display(
    combined_data.groupby("borough")[
        ["unique_complaints", "total_severe_injuries"]
    ].mean().round(2)
)
```

### Exploratory Data Analysis (EDA) Findings

We explored the relationship between NYC housing complaints and severe transportation injuries across five boroughs from 2022 to June 2026.

**Main findings:**
1. The Bronx recorded the highest average number of housing complaints.
2. Brooklyn recorded the highest average number of severe transportation injuries.
3. Staten Island recorded the lowest averages in both categories.
4. Brooklyn had the highest severe injury total in each year shown.
5. The scatter plot did not show a clear linear relationship between housing complaints and severe transportation injuries.

**Visualizations created:**
- Housing complaints by borough and year (line chart).
- Severe transportation injuries by borough and year (line chart).
- Housing complaints vs. severe transportation injuries (scatter plot).
- Severe transportation injuries by borough and year (heatmap).

**Limitation:** 2026 data covers only January to June and should not be directly compared with complete calendar years. The analysis identifies patterns and associations, not causal relationships.
